#Verify Protein Datasets

```python
!pip install biopython
from Bio import SeqIO

files = {
    "AGLA": "Agla2.0_prot.faa",
    "LDEC": "Ldec2.0_prot.faa",
    "TCAS": "Tcas5.2_prot.faa"
}

for species, filename in files.items():

    records = list(SeqIO.parse(filename, "fasta"))

    lengths = [len(record.seq) for record in records]

    print(f"\n{species}")
    print("Total proteins:", len(records))
    print("Shortest protein:", min(lengths))
    print("Longest protein:", max(lengths))
    print("Average length:", sum(lengths)/len(lengths))
```
