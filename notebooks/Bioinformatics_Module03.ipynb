####**TASK P7.4**. Ask Gemini how to <font color=crimson>```insert```</font> a new item at index 1 in the list

Here's how it works:

Syntax: list.insert(index, element)

index: The position where you want to insert the element. The element will be inserted before this index.
element: The item you want to insert into the list.
Behavior:

If index is 0, the element is inserted at the beginning of the list.
If index is greater than or equal to the length of the list, the element is appended to the end of the list.
If index is negative, it counts from the end of the list (e.g., -1 is one position from the end, -len(list) is the beginning). If the negative index is beyond the start of the list, it effectively inserts at the beginning.

````python
# On line #2 insert an item at index 1 of my_list
my_list.insert(1,"Apple")
print (my_list)
````

####**TASK BP21.2** <font color=orange> **Complete Rosalind #8  Translating RNA into Protein**</font>

````python
#An RNA string s corresponding to a strand of mRNA (of length at most 10 kbp)
#The protein string encoded by s
s=input('RNA Strand: ')
````

### RNA Codon Table

To translate RNA into protein, we use a codon table that maps each three-nucleotide codon to a specific amino acid. There are 64 possible codons, including start and stop codons.

````python
codon_table = {
    "UUU": "F", "UUC": "F", "UUA": "L", "UUG": "L",
    "CUU": "L", "CUC": "L", "CUA": "L", "CUG": "L",
    "AUU": "I", "AUC": "I", "AUA": "I", "AUG": "M",
    "GUU": "V", "GUC": "V", "GUA": "V", "GUG": "V",

    "UCU": "S", "UCC": "S", "UCA": "S", "UCG": "S",
    "CCU": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "ACU": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "GCU": "A", "GCC": "A", "GCA": "A", "GCG": "A",

    "UAU": "Y", "UAC": "Y", "UAA": "Stop", "UAG": "Stop",
    "CAU": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
    "AAU": "N", "AAC": "N", "AAA": "K", "AAG": "K",
    "GAU": "D", "GAC": "D", "GAA": "E", "GAG": "E",

    "UGU": "C", "UGC": "C", "UGA": "Stop", "UGG": "W",
    "CGU": "R", "CGC": "R", "CGA": "R", "CGG": "R",
    "AGU": "S", "AGC": "S", "AGA": "R", "AGG": "R",
    "GGU": "G", "GGC": "G", "GGA": "G", "GGG": "G"
}
````

````python
protein_string = ""
for i in range(0, len(s), 3):
    codon = s[i:i+3]
    if len(codon) == 3:
        amino_acid = codon_table.get(codon)
        if amino_acid == "Stop":
            break
        elif amino_acid:
            protein_string += amino_acid

print(protein_string)
````
